DATA ANALYSIS AND VISUALIZATION


HOW ARE SALES CHANING OVER TIME?

```python
monthly_sales=df.groupby(df['date'].dt.to_period('M'))['total_value'].sum()
monthly_sales.index=monthly_sales.index.astype(str)


#visualization--LINE CHART
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker='o',color='purple')
plt.title('Monthly sales trend')
plt.xlabel('Month')
plt.ylabel('Total sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

WHICH CATEGORY GENERATE THE. HIGHEST SALES AND PROFIT? (SOL BELOW)

```python
# formula 
# profit=totalvalue=quantity*cost_price

df['profit']=df['total_value']-(df['quantity']*df['cost_price'])
```

```python
df.head()

#last me profit column bann gya
```

```python
#analysis

category_perf=df.groupby('category').agg(total_sales=('total_value','sum'),total_profit=('profit','sum')).sort_values('total_sales',ascending=False)

category_perf
```

```python
#visualization -- categories -- barchart

x=np.arange(len(category_perf))
width= 0.35

plt.figure(figsize=(11,5))

plt.bar(x-width/2, category_perf['total_sales'],width=width,label='Total sales',color='blue')
plt.bar(x+width/2, category_perf['total_profit'],width=width,label='Total Profit',color='green')
 
plt.xticks(x,category_perf.index,rotation=90)
plt.ylabel('Amount')
plt.xlabel('Category')
plt.title('Category-wise sales vs profit')
plt.legend() 
plt.tight_layout()
plt.show()
```

Q3. WHICH PRODUCT ARE THE MOST PROFITABLE?


```python
product_perf=df.groupby('subcategory').agg(total_sales=('total_value','sum'), total_profit=('profit','sum'))

top_products=product_perf.sort_values('total_profit',ascending=False)

top10_profitable=top_products.head(10)
```

```python
#VISUALIZATION -- BAR CHART

plt.figure(figsize=(11,5))
plt.bar(top10_profitable.index,top10_profitable['total_profit'],color='red')
plt.xlabel('Product')
plt.ylabel('Total profit')
plt.title('Top 10 most profitable products')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
