HOW ARE SALES CHANING OVER TIME?

```python
monthly_sales=df.groupby(df['date'].dt.to_period('M'))['total_value'].sum()
monthly_sales.index=monthly_sales.index.astype(str)


#visualization--LINE CHART
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker='o',color='purple')
plt.title('Monthly sales trend')
plt.xlabel('Month')
plt.ylabel('Total sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

WHICH CATEGORY GENERATE THE. HIGHEST SALES AND PROFIT? (SOL BELOW)

```python
# formula 
# profit=totalvalue=quantity*cost_price

df['profit']=df['total_value']-(df['quantity']*df['cost_price'])
```

```python
df.head()

#last me profit column bann gya
```

```python
#analysis

category_perf=df.groupby('category').agg(total_sales=('total_value','sum'),total_profit=('profit','sum')).sort_values('total_sales',ascending=False)

category_perf
```

```python
#visualization -- categories -- barchart

x=np.arange(len(category_perf))
width= 0.35

plt.figure(figsize=(11,5))

plt.bar(x-width/2, category_perf['total_sales'],width=width,label='Total sales',color='blue')
plt.bar(x+width/2, category_perf['total_profit'],width=width,label='Total Profit',color='green')
 
plt.xticks(x,category_perf.index,rotation=90)
plt.ylabel('Amount')
plt.xlabel('Category')
plt.title('Category-wise sales vs profit')
plt.legend() 
plt.tight_layout()
plt.show()
```

Q3. WHICH PRODUCT ARE THE MOST PROFITABLE?


```python
product_perf=df.groupby('subcategory').agg(total_sales=('total_value','sum'), total_profit=('profit','sum'))

top_products=product_perf.sort_values('total_profit',ascending=False)

top10_profitable=top_products.head(10)
```

```python
#VISUALIZATION -- BAR CHART

plt.figure(figsize=(11,5))
plt.bar(top10_profitable.index,top10_profitable['total_profit'],color='red')
plt.xlabel('Product')
plt.ylabel('Total profit')
plt.title('Top 10 most profitable products')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# --- 1. BRAND PERFORMANCE ---
brand_perf = df.groupby('brand').agg(
    total_sales=('total_value', 'sum'),
    total_profit=('profit', 'sum'),
    units_sold=('quantity', 'sum')
)
brand_perf['profit_margin_%'] = (brand_perf['total_profit'] / brand_perf['total_sales']) * 100
brand_perf = brand_perf.sort_values(by='total_profit', ascending=False)

print("=== BRAND PERFORMANCE ===")
display(brand_perf.round(2))

# --- 2. TOP 10 MOST PROFITABLE PRODUCTS ---
product_perf = df.groupby(['sku_name', 'category', 'brand']).agg(
    total_sales=('total_value', 'sum'),
    total_profit=('profit', 'sum'),
    units_sold=('quantity', 'sum')
)
product_perf['profit_margin_%'] = (product_perf['total_profit'] / product_perf['total_sales']) * 100
top_products = product_perf.sort_values(by='total_profit', ascending=False).head(10)

print("\n=== TOP 10 MOST PROFITABLE PRODUCTS ===")
display(top_products.round(2))
```

```python


fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# --- Chart 1: Top 10 Most Profitable Products ---
top_10_plot = top_products.reset_index().sort_values(by='total_profit', ascending=True)

bars = axes[0].barh(
    top_10_plot['sku_name'], 
    top_10_plot['total_profit'] / 1e3, 
    color='#2b5c8f', 
    edgecolor='black',
    alpha=0.85
)
axes[0].set_title('Top 10 Most Profitable Products', fontsize=14, fontweight='bold', pad=12)
axes[0].set_xlabel('Total Profit (in Thousands AED / Currency)', fontsize=11)
axes[0].grid(axis='x', linestyle='--', alpha=0.5)

# Value annotations on bars
for bar in bars:
    width = bar.get_width()
    axes[0].text(width + 10, bar.get_y() + bar.get_height()/2, f'{width:.1f}K', va='center', fontsize=9, fontweight='semibold')

# --- Chart 2: Brand-wise Sales vs Profit ---
x = np.arange(len(brand_perf))
width = 0.35

axes[1].bar(x - width/2, brand_perf['total_sales'] / 1e6, width=width, label='Total Sales (M)', color='#4a90e2', edgecolor='black', alpha=0.85)
axes[1].bar(x + width/2, brand_perf['total_profit'] / 1e6, width=width, label='Total Profit (M)', color='#2e7d32', edgecolor='black', alpha=0.85)

axes[1].set_title('Brand-wise Sales vs Profit', fontsize=14, fontweight='bold', pad=12)
axes[1].set_xticks(x)
axes[1].set_xticklabels(brand_perf.index, fontsize=11)
axes[1].set_ylabel('Amount (in Millions)', fontsize=11)
axes[1].legend(frameon=True)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### Key Insights - Q3: Most Profitable Products & Brands

1. **Most Profitable Brand**:
   - **`Premium`** is the #1 most profitable brand, generating **~7.21 Million** in profit (out of 24.56M sales) with a **29.36%** profit margin.
   - Followed by **`Budget`** (4.27M profit) and **`BlueMart`** (3.37M profit).
   - In terms of profit margin percentage, **`BlueMart` (29.85%)** and **`Premium` (29.36%)** deliver the highest margins.

2. **Most Profitable Product**:
   - The single most profitable product is **`Electronics_Batteries_1035` (Brand: Premium)**, yielding **~768.5K profit** at a high margin of **39.04%**.
   - Second highest is **`Electronics_Mobile Accessories_1174` (Premium)** with **~690K profit** (37.27% margin).

3. **Category Trend**:
   - **All top 10 most profitable products belong to the `Electronics` category** (specifically Batteries, Cables, Mobile Accessories, and Chargers).
